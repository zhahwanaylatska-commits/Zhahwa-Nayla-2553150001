2. Samakan volume dulu (z-score, Pertemuan 2)

```python
X = w8[["lama_tinggal", "akomodasi_per_malam_juta"]].to_numpy(dtype=float)   # matriks 8 x 2

rata = X.mean(axis=0)            # rata-rata tiap kolom  -> [5.25, 0.78]
simp = X.std(axis=0, ddof=0)     # simpangan baku tiap kolom (populasi) -> [3.38, 0.52]
Z = (X - rata) / simp            # z-score: tiap kolom kini rata-rata 0, simpangan 1

print("rata-rata :", rata.round(2)); print("simpangan :", simp.round(3))
tabel_z = pd.DataFrame(Z.round(2), columns=["z_lama", "z_akomodasi"], index=w8.wisatawan)
tabel_z
```

3. K-Means dari nol — fungsi-fungsinya

```python
def jarak_ke_pusat(Z, pusat):
    """Mengembalikan matriks (n_titik x k): jarak Euclidean titik i ke pusat j.
    Z[:, None, :] - pusat[None, :, :] memakai broadcasting numpy supaya tidak perlu loop."""
    return np.sqrt(((Z[:, None, :] - pusat[None, :, :]) ** 2).sum(axis=2))

def bagi(Z, pusat):
    """Langkah 'bagi': label tiap titik = indeks pusat terdekat."""
    return jarak_ke_pusat(Z, pusat).argmin(axis=1)

def geser(Z, label, k):
    """Langkah 'geser': pusat baru = rata-rata anggota tiap kerumunan."""
    return np.vstack([Z[label == j].mean(axis=0) for j in range(k)])

def inertia(Z, pusat, label):
    """J = Σ ||z_i − pusat_{label_i}||²  (total jarak kuadrat ke pusat sendiri)."""
    return ((Z - pusat[label]) ** 2).sum()
```

4. Jalankan dua putaran dengan tangan (slide 6)

```python
k = 2
pusat = Z[[0, 4]].copy()            # pusat awal: W1 (indeks 0) dan W5 (indeks 4)
nama = np.array(["A", "B"])

for putaran in range(1, 6):
    D = jarak_ke_pusat(Z, pusat)                      # jarak tiap titik ke pusat A dan B
    label = D.argmin(axis=1)                          # BAGI
    J = inertia(Z, pusat, label)
    print(f"--- Putaran {putaran} --- (J = {J:.2f})")
    tampil = pd.DataFrame({"d ke A": D[:, 0].round(2), "d ke B": D[:, 1].round(2), "ikut": nama[label]}, index=w8.wisatawan)
    print(tampil.to_string())
    pusat_baru = geser(Z, label, k)                   # GESER
    print("pusat baru (skala z):", pusat_baru.round(2).tolist())
    print("pusat baru (satuan asli):", (pusat_baru * simp + rata).round(2).tolist(), "\n")
    if np.allclose(pusat_baru, pusat):                # tidak ada pusat yang bergerak -> BERHENTI
        print(f"Pusat tidak bergerak lagi -> konvergen pada putaran {putaran}.")
        break
    pusat = pusat_baru
```

5. Eksperimen: apa yang terjadi TANPA z-score?

```python
X_ribu = X.copy(); X_ribu[:, 1] *= 1000             # akomodasi dalam ribu rupiah: 1.200, 1.500, ...
pusat_m = X_ribu[[0, 4]].copy()
for _ in range(10):
    lab_m = bagi(X_ribu, pusat_m); baru = geser(X_ribu, lab_m, 2)
    if np.allclose(baru, pusat_m): break
    pusat_m = baru
D_m = jarak_ke_pusat(X_ribu, pusat_m)
print("Jarak W1 ke pusat A dan B tanpa skala:", D_m[0].round(1), " <- angka ratusan: semuanya dari rupiah")
print("Label tanpa skala  :", nama[lab_m])
print("Label dengan z-score:", nama[label])
print("Di 8 titik ini hasilnya kebetulan sama (dua kolom 'setuju'); di 1.200 wisatawan notebook 02, hasilnya BERBEDA jauh.")
```

6. Bonus: silhouette dihitung tangan (slide 11)

```python
def silhouette_manual(Z, label):
    n = len(Z); D = np.sqrt(((Z[:, None, :] - Z[None, :, :]) ** 2).sum(axis=2))   # jarak antar semua titik
    s = np.zeros(n)
    for i in range(n):
        sendiri = (label == label[i]); sendiri[i] = False             # anggota kerumunan sendiri, kecuali diri sendiri
        a = D[i, sendiri].mean()
        b = min(D[i, label == j].mean() for j in set(label) if j != label[i])   # kerumunan lain terdekat
        s[i] = (b - a) / max(a, b)
    return s

s = silhouette_manual(Z, label)
print(pd.DataFrame({"s": s.round(2)}, index=w8.wisatawan).T)
print("Skor silhouette (rata-rata):", s.mean().round(3))

from sklearn.metrics import silhouette_score          # cek dengan scikit-learn: harus sama
print("scikit-learn               :", round(silhouette_score(Z, label), 3))
```

02 · K-Means pada 1.200 Wisatawan: Kode #1 & Kode #2

```python
# ============================================================
# PERSIAPAN — jalankan sel ini dulu
# ============================================================
import os, io, urllib.request
import numpy as np, pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.width", 160); pd.set_option("display.max_columns", 30)
plt.rcParams["figure.dpi"] = 110

# Folder data. Kalau notebook dibuka dari folder "kode/", data ada di "data/".
# Di Colab: upload CSV ke panel Files (ikon folder di kiri) ke folder "data/",
# ATAU biarkan fungsi muat() mengunduh/menyusun ulang datanya.
DATA_DIR = "data"
os.makedirs(DATA_DIR, exist_ok=True)

def muat(nama_file, url_cadangan=None, pembuat=None):
    """Memuat CSV dari data/<nama_file>.
    - Kalau file tidak ada dan ada url_cadangan  -> unduh dari internet lalu simpan.
    - Kalau file tidak ada dan ada pembuat()     -> buat ulang data (sintetis, seed tetap) lalu simpan.
    Dengan begitu notebook tetap jalan di Colab walau folder data/ kosong."""
    path = os.path.join(DATA_DIR, nama_file)
    if os.path.exists(path):
        print(f"[muat] membaca {path}")
        return pd.read_csv(path)
    if url_cadangan is not None:
        print(f"[muat] {path} tidak ada -> mengunduh dari {url_cadangan}")
        urllib.request.urlretrieve(url_cadangan, path)
        return pd.read_csv(path)
    if pembuat is not None:
        print(f"[muat] {path} tidak ada -> membuat ulang data sintetis")
        df = pembuat(); df.to_csv(path, index=False)
        return df
    raise FileNotFoundError(path)

print("numpy", np.__version__, "| pandas", pd.__version__)
```

```python
def buat_wisatawan_1200():
    """Data sintetis 1.200 wisatawan resort Bali (seed 6) — SAMA dengan slide.
    Empat 'tipe' tersembunyi hanya dipakai untuk MEMBUAT data; kolomnya tidak disimpan,
    karena di dunia nyata hotel memang tidak punya label itu."""
    rng = np.random.default_rng(6); n = 1200
    tipe = rng.choice(4, n, p=[0.30, 0.30, 0.22, 0.18])      # 0 backpacker, 1 keluarga, 2 bisnis, 3 premium
    mu = {0: [8, 250, 90, 2, 10, 1.3], 1: [4, 900, 350, 4, 45, 3.8],
          2: [2, 1200, 200, 0.5, 7, 1.1], 3: [3, 2800, 700, 5, 60, 2.0]}
    sd = {0: [2.5, 80, 30, 1, 6, 0.5], 1: [1.2, 250, 100, 1.5, 15, 1.0],
          2: [0.8, 300, 70, 0.7, 4, 0.3], 3: [1, 600, 200, 1.5, 20, 0.2]}
    X = np.zeros((n, 6))
    for s in range(4):
        m = tipe == s
        X[m] = rng.normal(mu[s], sd[s], size=(m.sum(), 6))
    df = pd.DataFrame(X, columns=["lama_tinggal", "belanja_akomodasi_per_malam", "belanja_kuliner_per_hari",
                                  "jumlah_aktivitas", "lead_time_hari", "jumlah_orang"])
    # pembulatan & batas wajar per kolom (persis seperti pembuatan angka di slide)
    df["lama_tinggal"] = df.lama_tinggal.clip(1, 20).round().astype(int)
    df["belanja_akomodasi_per_malam"] = df.belanja_akomodasi_per_malam.clip(100, 6000).round(-1)
    df["belanja_kuliner_per_hari"] = df.belanja_kuliner_per_hari.clip(30, 2000).round(-1)
    df["jumlah_aktivitas"] = df.jumlah_aktivitas.clip(0, 12).round().astype(int)
    df["lead_time_hari"] = df.lead_time_hari.clip(0, 180).round().astype(int)
    df["jumlah_orang"] = df.jumlah_orang.clip(1, 8).round().astype(int)
    df.insert(0, "id_wisatawan", [f"T{i:04d}" for i in range(1, n + 1)])
    return df
```

```python
wisatawan = muat("wisatawan_1200_sintetis.csv", pembuat=buat_wisatawan_1200)
FITUR = ["lama_tinggal", "belanja_akomodasi_per_malam", "belanja_kuliner_per_hari",
         "jumlah_aktivitas", "lead_time_hari", "jumlah_orang"]
print(wisatawan.shape)
wisatawan.head()
```

2. Kode #1 — samakan volume, cari 4 kerumunan

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

skala = StandardScaler().fit(wisatawan[FITUR])     # pelajari rata-rata & simpangan tiap kolom
Z = skala.transform(wisatawan[FITUR])              # z-score: matriks 1200 x 6

km = KMeans(n_clusters=4, n_init=10, random_state=0).fit(Z)   # hanya Z, tidak ada y!
wisatawan["cluster"] = km.labels_                              # 0..3 — hanya nomor, belum ada arti

print("inertia J :", round(km.inertia_, 1))                    # acuan seed 6: 1794,7
print("ukuran    :", wisatawan.cluster.value_counts().sort_index().to_dict())
```

5. Kode #2 — berapa k? Elbow + silhouette (slide 10–11)

```python
from sklearn.metrics import silhouette_score

hasil = []
for k in range(2, 9):                                    # k = 1 tidak punya silhouette (error) -> mulai dari 2
    km_k = KMeans(n_clusters=k, n_init=10, random_state=0).fit(Z)
    hasil.append({"k": k, "inertia": round(km_k.inertia_), "silhouette": round(silhouette_score(Z, km_k.labels_), 3)})
hasil = pd.DataFrame(hasil).set_index("k")
print(hasil.T)

fig, ax = plt.subplots(1, 2, figsize=(10, 3.6))
ax[0].plot(hasil.index, hasil.inertia, "o-", color="#F7941E"); ax[0].set_title("Elbow: inertia vs k"); ax[0].set_xlabel("k")
ax[1].plot(hasil.index, hasil.silhouette, "o-", color="#C97407"); ax[1].set_title("Silhouette vs k"); ax[1].set_xlabel("k")
for a in ax: a.axvline(4, color="gray", ls="--", lw=1)
plt.tight_layout(); plt.show()

turun = hasil.inertia.pct_change().abs().mul(100).round(0)
print("Persentase penurunan inertia dari k sebelumnya:", turun.dropna().astype(int).to_dict())
```

```python
# ============================================================
# PERSIAPAN — jalankan sel ini dulu
# ============================================================
import os, io, urllib.request
import numpy as np, pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.width", 160); pd.set_option("display.max_columns", 30)
plt.rcParams["figure.dpi"] = 110

# Folder data. Kalau notebook dibuka dari folder "kode/", data ada di "data/".
# Di Colab: upload CSV ke panel Files (ikon folder di kiri) ke folder "data/",
# ATAU biarkan fungsi muat() mengunduh/menyusun ulang datanya.
DATA_DIR = "data"
os.makedirs(DATA_DIR, exist_ok=True)

def muat(nama_file, url_cadangan=None, pembuat=None):
    """Memuat CSV dari data/<nama_file>.
    - Kalau file tidak ada dan ada url_cadangan  -> unduh dari internet lalu simpan.
    - Kalau file tidak ada dan ada pembuat()     -> buat ulang data (sintetis, seed tetap) lalu simpan.
    Dengan begitu notebook tetap jalan di Colab walau folder data/ kosong."""
    path = os.path.join(DATA_DIR, nama_file)
    if os.path.exists(path):
        print(f"[muat] membaca {path}")
        return pd.read_csv(path)
    if url_cadangan is not None:
        print(f"[muat] {path} tidak ada -> mengunduh dari {url_cadangan}")
        urllib.request.urlretrieve(url_cadangan, path)
        return pd.read_csv(path)
    if pembuat is not None:
        print(f"[muat] {path} tidak ada -> membuat ulang data sintetis")
        df = pembuat(); df.to_csv(path, index=False)
        return df
    raise FileNotFoundError(path)

print("numpy", np.__version__, "| pandas", pd.__version__)
```

```python
def buat_survei_600():
    """Survei kepuasan 600 tamu, 10 pertanyaan skala 1-5 (seed 66) — SAMA dengan slide.
    Dibangun dari TIGA 'perasaan' tersembunyi: fisik, layanan manusia, nilai uang.
    PCA nanti diharapkan 'menemukan' ketiganya tanpa pernah diberi tahu."""
    rng = np.random.default_rng(66); m = 600
    f_fisik, f_layanan, f_nilai = rng.normal(0, 1, (3, m))
    bobot = {"kebersihan_kamar": (0.9, 0.1, 0), "kenyamanan_kasur": (0.85, 0, 0.1), "kebersihan_kolam": (0.8, 0.1, 0),
             "kecepatan_wifi": (0.6, 0, 0.2), "keramahan_staf": (0.1, 0.9, 0), "kecepatan_checkin": (0, 0.85, 0.1),
             "kualitas_sarapan": (0.3, 0.6, 0.1), "info_tur_jelas": (0, 0.8, 0.1), "harga_sepadan": (0.2, 0.1, 0.9),
             "variasi_menu": (0.3, 0.3, 0.6)}
    S = {}
    for q, (a, b, c) in bobot.items():
        v = 3.6 + 0.7 * (a * f_fisik + b * f_layanan + c * f_nilai) + rng.normal(0, 0.35, m)
        S[q] = v.clip(1, 5).round().astype(int)
    df = pd.DataFrame(S); df.insert(0, "id_responden", [f"R{i:03d}" for i in range(1, m + 1)])
    return df
```

```python
survei = muat("survei_kepuasan_600_sintetis.csv", pembuat=buat_survei_600)
Q = [c for c in survei.columns if c != "id_responden"]       # 10 pertanyaan
print(survei.shape); survei.head()
```

2. PCA dari nol — lima langkah slide 15

```python
X = survei[Q].to_numpy(dtype=float)
n = len(X)

# 1) samakan volume (ddof=0 agar sama dengan StandardScaler scikit-learn)
Zs = (X - X.mean(axis=0)) / X.std(axis=0, ddof=0)

# 2) matriks kovarians 10 x 10 (pembagi n-1 seperti scikit-learn)
S = Zs.T @ Zs / (n - 1)

# 3) nilai & vektor eigen; eigh untuk matriks simetris, hasil urut NAIK -> balik jadi TURUN
lam, W = np.linalg.eigh(S)
urut = np.argsort(lam)[::-1]; lam, W = lam[urut], W[:, urut]

# 4) porsi varians yang dijelaskan
porsi = lam / lam.sum()
print("porsi tiap sumbu   :", porsi.round(3))
print("kumulatif          :", porsi.cumsum().round(2))

# 5) skor tiap tamu pada tiga sumbu pertama = proyeksi (dot product)
skor_manual = Zs @ W[:, :3]
print("skor 3 tamu pertama:\n", skor_manual[:3].round(2))
```

3. Cek dengan scikit-learn — angkanya harus sama

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

skala = StandardScaler().fit(survei[Q]); Zs_sk = skala.transform(survei[Q])
pca = PCA().fit(Zs_sk)                                   # semua 10 komponen
print("porsi (sklearn)  :", pca.explained_variance_ratio_.round(3))
print("sama dengan manual?", np.allclose(pca.explained_variance_ratio_, porsi))

# bandingkan bobot PC1 (abaikan tanda)
print("bobot PC1 manual :", np.abs(W[:, 0]).round(2))
print("bobot PC1 sklearn:", np.abs(pca.components_[0]).round(2))
```

4. Scree plot — berapa sumbu yang cukup? (slide 17)

```python
porsi_sk = pca.explained_variance_ratio_ * 100
fig, ax = plt.subplots(figsize=(7, 3.8))
ax.bar(range(1, 11), porsi_sk, color="#F7941E")
ax.plot(range(1, 11), porsi_sk.cumsum(), "o-", color="#1C1C1C", label="kumulatif")
for i, v in enumerate(porsi_sk): ax.text(i + 1, v + 1, f"{v:.1f}", ha="center", fontsize=8)
ax.axhline(70, color="gray", ls="--", lw=1); ax.set_xticks(range(1, 11)); ax.set_xticklabels([f"PC{i}" for i in range(1, 11)])
ax.set_ylabel("% varians"); ax.legend(); ax.set_title("Scree plot survei kepuasan"); plt.show()
print("Tiga sumbu menangkap", porsi_sk[:3].sum().round(1), "% perbedaan antar tamu")
```

```python
def buat_wisatawan_1200():
    """Data sintetis 1.200 wisatawan resort Bali (seed 6) — SAMA dengan slide.
    Empat 'tipe' tersembunyi hanya dipakai untuk MEMBUAT data; kolomnya tidak disimpan,
    karena di dunia nyata hotel memang tidak punya label itu."""
    rng = np.random.default_rng(6); n = 1200
    tipe = rng.choice(4, n, p=[0.30, 0.30, 0.22, 0.18])      # 0 backpacker, 1 keluarga, 2 bisnis, 3 premium
    mu = {0: [8, 250, 90, 2, 10, 1.3], 1: [4, 900, 350, 4, 45, 3.8],
          2: [2, 1200, 200, 0.5, 7, 1.1], 3: [3, 2800, 700, 5, 60, 2.0]}
    sd = {0: [2.5, 80, 30, 1, 6, 0.5], 1: [1.2, 250, 100, 1.5, 15, 1.0],
          2: [0.8, 300, 70, 0.7, 4, 0.3], 3: [1, 600, 200, 1.5, 20, 0.2]}
    X = np.zeros((n, 6))
    for s in range(4):
        m = tipe == s
        X[m] = rng.normal(mu[s], sd[s], size=(m.sum(), 6))
    df = pd.DataFrame(X, columns=["lama_tinggal", "belanja_akomodasi_per_malam", "belanja_kuliner_per_hari",
                                  "jumlah_aktivitas", "lead_time_hari", "jumlah_orang"])
    # pembulatan & batas wajar per kolom (persis seperti pembuatan angka di slide)
    df["lama_tinggal"] = df.lama_tinggal.clip(1, 20).round().astype(int)
    df["belanja_akomodasi_per_malam"] = df.belanja_akomodasi_per_malam.clip(100, 6000).round(-1)
    df["belanja_kuliner_per_hari"] = df.belanja_kuliner_per_hari.clip(30, 2000).round(-1)
    df["jumlah_aktivitas"] = df.jumlah_aktivitas.clip(0, 12).round().astype(int)
    df["lead_time_hari"] = df.lead_time_hari.clip(0, 180).round().astype(int)
    df["jumlah_orang"] = df.jumlah_orang.clip(1, 8).round().astype(int)
    df.insert(0, "id_wisatawan", [f"T{i:04d}" for i in range(1, n + 1)])
    return df
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA

wisatawan = muat("wisatawan_1200_sintetis.csv", pembuat=buat_wisatawan_1200)
FITUR = ["lama_tinggal", "belanja_akomodasi_per_malam", "belanja_kuliner_per_hari", "jumlah_aktivitas", "lead_time_hari", "jumlah_orang"]
Z = StandardScaler().fit_transform(wisatawan[FITUR])     # samakan volume
print(wisatawan.shape); wisatawan.head(3)
```

Bagian A · Berapa kerumunan? (10 menit)

```python
hasil = []
for k in range(2, 9):
    km_z = KMeans(n_clusters=k, n_init=10, random_state=0).fit(Z)                   # dengan skala
    km_m = KMeans(n_clusters=k, n_init=10, random_state=0).fit(wisatawan[FITUR])    # data mentah
    hasil.append({"k": k, "inertia_Z": round(km_z.inertia_),
                  "silhouette_Z": round(silhouette_score(Z, km_z.labels_), 3),
                  "silhouette_mentah": round(silhouette_score(wisatawan[FITUR], km_m.labels_), 3)})
hasil = pd.DataFrame(hasil).set_index("k"); print(hasil)
hasil[["silhouette_Z", "silhouette_mentah"]].plot(marker="o", figsize=(6, 3)); plt.show()

k_pilihan = 4
# ALASAN: siku inertia di k=4 (turun 35% dari 3->4, hanya 9% dari 4->5); silhouette Z puncak 0,477 di k=4.
# Silhouette data MENTAH justru lebih tinggi -> kerumunan 'harga' yang bulat rapi tapi tidak berguna (keluarga & pebisnis melebur).
```

```python
# Penamaan OTOMATIS hanya sebagai BANTUAN: dua kategori dengan selisih tertinggi per cluster.
# Nama akhir tetap keputusan manusia (lihat slide 8).
for c in range(K):
    atas = selisih.loc[c].sort_values(ascending=False)
    print(f"cluster {c} (n={profil.n[c]}): paling suka {atas.index[0]} (+{atas.iloc[0]:.2f}), {atas.index[1]} (+{atas.iloc[1]:.2f}) | paling kurang {atas.index[-1]} ({atas.iloc[-1]:.2f})")
```

06 · Data Nyata — PCA pada Survei Kepuasan Penumpang Maskapai (Airline Passenger Satisfaction)

```python
# ============================================================
# PERSIAPAN — jalankan sel ini dulu
# ============================================================
import os, io, urllib.request
import numpy as np, pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.width", 160); pd.set_option("display.max_columns", 30)
plt.rcParams["figure.dpi"] = 110

# Folder data. Kalau notebook dibuka dari folder "kode/", data ada di "data/".
# Di Colab: upload CSV ke panel Files (ikon folder di kiri) ke folder "data/",
# ATAU biarkan fungsi muat() mengunduh/menyusun ulang datanya.
DATA_DIR = "data"
os.makedirs(DATA_DIR, exist_ok=True)

def muat(nama_file, url_cadangan=None, pembuat=None):
    """Memuat CSV dari data/<nama_file>.
    - Kalau file tidak ada dan ada url_cadangan  -> unduh dari internet lalu simpan.
    - Kalau file tidak ada dan ada pembuat()     -> buat ulang data (sintetis, seed tetap) lalu simpan.
    Dengan begitu notebook tetap jalan di Colab walau folder data/ kosong."""
    path = os.path.join(DATA_DIR, nama_file)
    if os.path.exists(path):
        print(f"[muat] membaca {path}")
        return pd.read_csv(path)
    if url_cadangan is not None:
        print(f"[muat] {path} tidak ada -> mengunduh dari {url_cadangan}")
        urllib.request.urlretrieve(url_cadangan, path)
        return pd.read_csv(path)
    if pembuat is not None:
        print(f"[muat] {path} tidak ada -> membuat ulang data sintetis")
        df = pembuat(); df.to_csv(path, index=False)
        return df
    raise FileNotFoundError(path)

print("numpy", np.__version__, "| pandas", pd.__version__)
```

```python
URL_AIR = "https://raw.githubusercontent.com/zeyadmedhat/Airline-Passenger-Satisfaction/main/data/train.csv"
def unduh_sampel():
    df = pd.read_csv(URL_AIR).drop(columns=["Unnamed: 0"], errors="ignore")
    return df.sample(20000, random_state=6).sort_values("id")          # sampel 20.000 supaya ringan di Colab
air = muat("airline_passenger_satisfaction_20k.csv", pembuat=unduh_sampel)

LAYANAN = ["Inflight wifi service", "Departure/Arrival time convenient", "Ease of Online booking", "Gate location",
           "Food and drink", "Online boarding", "Seat comfort", "Inflight entertainment", "On-board service",
           "Leg room service", "Baggage handling", "Checkin service", "Inflight service", "Cleanliness"]
print(air.shape); air[LAYANAN].describe().round(2).T[["mean", "std", "min", "max"]]
```

1. Bersihkan: nilai 0 = "tidak berlaku"

```python
X = air[LAYANAN].replace(0, np.nan)
print("persentase nilai 0 per pertanyaan:"); print((X.isna().mean() * 100).round(1).sort_values(ascending=False).head(6))
X = X.fillna(X.median())
```

2. Korelasi → PCA → scree

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

skala = StandardScaler().fit(X); Zs = skala.transform(X)
pca = PCA().fit(Zs)
porsi = pca.explained_variance_ratio_ * 100
print("porsi   :", porsi.round(1)); print("kumulatif:", porsi.cumsum().round(1))

fig, ax = plt.subplots(figsize=(8, 3.6))
ax.bar(range(1, 15), porsi, color="#F7941E"); ax.plot(range(1, 15), porsi.cumsum(), "o-", color="black", label="kumulatif")
ax.axhline(70, color="gray", ls="--", lw=1); ax.set_xticks(range(1, 15)); ax.set_xlabel("komponen"); ax.set_ylabel("% varians"); ax.legend(); plt.show()
```

3. Membaca bobot: tema apa yang "ditemukan" PCA?

```python
NK = 3                                                      # jumlah komponen yang dipertahankan (lihat scree)
bobot = pd.DataFrame(pca.components_[:NK].T, index=LAYANAN, columns=[f"PC{i+1}" for i in range(NK)]).round(2)
def sorot(v): return "font-weight: bold; background-color: #FDF0DD" if abs(v) >= 0.3 else ""
try: display(bobot.style.map(sorot))
except Exception: print(bobot)

for pc in bobot.columns:
    print(pc, "| positif:", bobot[pc][bobot[pc] >= 0.3].index.tolist(), "| negatif:", bobot[pc][bobot[pc] <= -0.3].index.tolist())
```

```python
nama_indeks = {"PC1": "Indeks 1 (isi nama)", "PC2": "Indeks 2 (isi nama)", "PC3": "Indeks 3 (isi nama)"}
pcaK = PCA(n_components=NK).fit(Zs)                        # bekukan bobot
skor = pd.DataFrame(pcaK.transform(Zs), columns=list(nama_indeks.values()), index=air.index)

# Pemeriksaan akal sehat: indeks vs label kepuasan (label TIDAK dipakai saat membuat indeks)
cek = skor.groupby(air["satisfaction"]).mean().round(2); print(cek)
print("\nKalau indeks 1 jauh lebih tinggi pada penumpang 'satisfied', arah sumbunya 'positif = baik'; kalau sebaliknya, kalikan -1 dan beri nama sesuai arah.")
```

4. Segmen kepuasan: K-Means pada skor PCA (bekal Tugas B)

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
S = skor.to_numpy()
for k in (2, 3, 4, 5):
    kmk = KMeans(n_clusters=k, n_init=10, random_state=0).fit(S)
    print(f"k={k}: silhouette {silhouette_score(S, kmk.labels_, sample_size=5000, random_state=0):.3f}")
km = KMeans(n_clusters=4, n_init=10, random_state=0).fit(S)
seg = skor.assign(segmen=km.labels_).groupby("segmen").mean().round(2)
seg["n"] = np.bincount(km.labels_)
seg["% satisfied"] = (pd.crosstab(km.labels_, air["satisfaction"], normalize="index")["satisfied"] * 100).round(0)
seg["% business travel"] = (pd.crosstab(km.labels_, air["Type of Travel"], normalize="index")["Business travel"] * 100).round(0)
print(seg)
```

```python
URL_HB = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2020/2020-02-11/hotels.csv"
hb = muat("hotel_bookings.csv", url_cadangan=URL_HB)
print(hb.shape); hb.head(3)
```

```python
d = hb[(hb.hotel == "Resort Hotel") & (hb.is_canceled == 0)].copy()
d["total_malam"] = d.stays_in_weekend_nights + d.stays_in_week_nights
d["tamu"] = d.adults + d.children.fillna(0) + d.babies
FITUR = ["lead_time", "total_malam", "tamu", "adr", "total_of_special_requests", "previous_bookings_not_canceled", "booking_changes"]
print("Resort, tidak batal:", len(d)); print(d[FITUR].describe().round(1).T)
```

```python
sebelum = len(d)
d = d[(d.adr > 0) & (d.adr < 1000) & (d.total_malam > 0) & (d.tamu > 0)]
print(f"dibuang {sebelum - len(d)} baris -> tersisa {len(d)}")
d[FITUR].hist(figsize=(11, 6), bins=30, color="#F7941E"); plt.tight_layout(); plt.show()
```

3. Scaling → loop k

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

skala = StandardScaler().fit(d[FITUR]); Z = skala.transform(d[FITUR])
hasil = []
for k in range(2, 8):
    kmk = KMeans(n_clusters=k, n_init=10, random_state=0).fit(Z)
    hasil.append({"k": k, "inertia": round(kmk.inertia_), "silhouette": round(silhouette_score(Z, kmk.labels_, sample_size=5000, random_state=0), 3)})
hasil = pd.DataFrame(hasil).set_index("k"); print(hasil.T)
fig, ax = plt.subplots(1, 2, figsize=(10, 3.4)); ax[0].plot(hasil.index, hasil.inertia, "o-", color="#F7941E"); ax[0].set_title("inertia")
ax[1].plot(hasil.index, hasil.silhouette, "o-", color="#C97407"); ax[1].set_title("silhouette"); plt.show()
```

```python
K = 4
km = KMeans(n_clusters=K, n_init=10, random_state=0).fit(Z)
d["cluster"] = km.labels_
profil = d.groupby("cluster")[FITUR].mean().round(1); profil["n"] = d.cluster.value_counts().sort_index()
print(profil)

# Nama berdasarkan CIRI (bukan nomor): urutan nomor bisa berbeda di komputer lain
nama = {}
nama[profil.previous_bookings_not_canceled.idxmax()] = "Tamu setia berulang (kecil tapi bernilai)"
nama[profil.lead_time.idxmax()] = "Perencana liburan panjang"
nama[profil.adr.idxmax()] = "Keluarga/rombongan tarif tinggi"
for c in profil.index:
    nama.setdefault(c, "Pelancong singkat spontan")
profil["nama"] = pd.Series(nama); profil
```
